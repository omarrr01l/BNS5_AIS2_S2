17. Replace the last value of each tuple in a list

- Sample list:
- [(10, 20, 40), (40, 50, 60), (70, 80, 90)]
- Expected output:
- [(10, 20, 100), (40, 50, 100), (70, 80, 100)]

```python
data = [(10, 20, 40), (40, 50, 60), (70, 80, 90)]

new_list = []

for item in data:
    new_item = list(item)
    new_item[-1] = 100
    new_item = tuple(new_item)
    new_list.append(new_item)
print(new_list)
```
